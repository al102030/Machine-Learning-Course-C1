# Line Graph

```python
from matplotlib import pyplot as plt
import numpy as np
import random
```

```python
ages_x = [x for x in range(25, 36)]

dev_y = sorted([random.randint(30000, 80000) for _ in range(11)])

dev_y
```

```python
plt.plot(ages_x, dev_y, color="r",linestyle="--", marker=".", label="All Devs")

plt.style.use('fivethirtyeight')

py_dev_x = [x for x in range(25, 36)]

py_dev_y = sorted([random.randint(30000, 80000) for _ in range(11)])

plt.plot(ages_x, py_dev_y, "b", marker="o", label="Python")

js_dev_y = sorted([random.randint(30000, 80000) for _ in range(11)])

plt.plot(ages_x, js_dev_y, color="#00ad3b", label="JavaScript", linewidth=3)

plt.title("Median salary (USD) by age")
plt.xlabel("Ages")
plt.ylabel("Median salary (USD)")

# plt.grid()
# plt.legend(['All Devs', 'Python'])
plt.legend()

plt.show()
```

# Bar Graph

```python
ages_x = [x for x in range(25, 36)]

x_indexes = np.arange(len(ages_x))
width = 0.25

dev_y = [random.randint(30000, 80000) for _ in range(11)]

plt.bar(x_indexes - width, dev_y, width=width, color="r", label="All Devs")


py_dev_y = [random.randint(30000, 80000) for _ in range(11)]

plt.bar(x_indexes, py_dev_y, width=width, color="b", label="Python")

js_dev_y = sorted([random.randint(30000, 80000) for _ in range(11)])

plt.bar(x_indexes + width, js_dev_y, width=width, color="y", label="JavaScript")




plt.title("Median salary (USD) by age")
plt.xlabel("Ages")
plt.ylabel("Median salary (USD)")

# plt.grid()
# plt.legend(['All Devs', 'Python'])
plt.legend()

plt.xticks(ticks=x_indexes, labels=ages_x)

plt.show()
```

```python
import pandas as pd
from collections import Counter

df = pd.read_csv("data.csv")

language_counter = Counter()

for languages in df["LanguagesWorkedWith"]:
    language_counter.update(languages.split(";"))

languages = []
popularity = []

for item in language_counter.most_common(15):
    languages.append(item[0])
    popularity.append(item[1])

print(languages)
print(popularity)
```

```python
languages.reverse()
popularity.reverse()


plt.barh(languages, popularity)
plt.title("Most popular Languages")
plt.ylabel("Programming Languages")
plt.xlabel("Number of People who Use")

plt.tight_layout()

plt.show()
```
